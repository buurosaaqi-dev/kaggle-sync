```python
import time, datetime, shutil, subprocess
print('HEARTBEAT START', datetime.datetime.now(datetime.timezone.utc).isoformat(), flush=True)
if shutil.which('nvidia-smi'):
    print(subprocess.run(['nvidia-smi','-L'],capture_output=True,text=True).stdout, flush=True)
else:
    print('no nvidia-smi (cpu session)', flush=True)
for i in range(60):
    print(f'HB {i} {datetime.datetime.now(datetime.timezone.utc).isoformat()}', flush=True)
    time.sleep(20)
print('HEARTBEAT DONE', flush=True)
```
