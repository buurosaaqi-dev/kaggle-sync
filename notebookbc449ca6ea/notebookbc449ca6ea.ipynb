````python
import os
import sys

os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

%pip install -q -U transformers accelerate bitsandbytes fastapi uvicorn pydantic
!which cloudflared > /dev/null || (wget -nc -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64.deb && dpkg -i cloudflared-linux-amd64.deb > /dev/null 2>&1)

import gc
import json
import re
import subprocess
import threading
import time
import uuid

import torch
import uvicorn
from fastapi import FastAPI, Request
from fastapi.responses import JSONResponse, StreamingResponse
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    StoppingCriteria,
    StoppingCriteriaList,
    TextIteratorStreamer,
)

assert torch.cuda.device_count() == 2, "This notebook needs the GPU T4 x2 accelerator."
model_id = "WhiteRabbitNeo/WhiteRabbitNeo-13B-v1"
served_name = "WhiteRabbitNeo-13B"
print("Loading", model_id, "on", torch.cuda.device_count(), "GPUs", flush=True)

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)
tokenizer = AutoTokenizer.from_pretrained(model_id)
tokenizer.truncation_side = "left"
try:
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        quantization_config=bnb_config,
        device_map={"": 0},
        max_memory={0: "13GiB"},
    )
    model.eval()
    with torch.inference_mode():
        check_inputs = tokenizer(
            "SYSTEM: " + ("Read the instructions and answer clearly. " * 400) + "\nUSER: Say ready.\nASSISTANT:",
            return_tensors="pt", truncation=True, max_length=2048,
        ).to(model.device)
        check_output = model.generate(
            **check_inputs, max_new_tokens=16, do_sample=False,
            pad_token_id=tokenizer.eos_token_id,
        )
        del check_inputs, check_output
except (ValueError, RuntimeError, torch.OutOfMemoryError) as exc:
    print("Single GPU load failed; using both T4 GPUs:", str(exc)[:300], flush=True)
    if "model" in globals():
        del model
    gc.collect()
    torch.cuda.empty_cache()
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        quantization_config=bnb_config,
        device_map="auto",
        max_memory={0: "6GiB", 1: "6GiB"},
    )
model.eval()
print("Model device map:", getattr(model, "hf_device_map", {"all": str(model.device)}), flush=True)

app = FastAPI()
generation_lock = threading.Lock()


class StopOnDialogue(StoppingCriteria):
    def __init__(self, prompt_tokens):
        self.prompt_tokens = prompt_tokens

    def __call__(self, input_ids, scores, **kwargs):
        if input_ids.shape[1] <= self.prompt_tokens:
            return False
        tail = tokenizer.decode(input_ids[0, -14:], skip_special_tokens=True)
        return bool(re.search(r"\n(?:USER|SYSTEM|ASSISTANT):\s*$", tail))


def clean_dialogue_tail(text):
    return re.split(r"\n(?:USER|SYSTEM|ASSISTANT):", text, maxsplit=1)[0].strip()


def build_inputs(prompt, latest_user):
    request_tokens = tokenizer(latest_user, add_special_tokens=False)["input_ids"][:192]
    request_excerpt = tokenizer.decode(request_tokens, skip_special_tokens=True)
    guidance = (
        "SYSTEM: You are WhiteRabbitNeo. Answer the latest USER request directly. "
        "Follow its requested language, length and format. Do not copy earlier replies, "
        "repeat introductions or timestamps, or invent another USER turn. "
        "Give a complete answer when details are requested. "
        "If the context requires Agent Zero tool JSON, preserve that format.\n"
        f"Latest USER request: {request_excerpt}\nCONTEXT:\n"
    )
    prefix = tokenizer(guidance, return_tensors="pt")["input_ids"][0]
    history = tokenizer(prompt, add_special_tokens=False, return_tensors="pt")["input_ids"][0]
    ids = torch.cat((prefix, history[-(2048 - prefix.numel()):])).unsqueeze(0).to(model.device)
    return {"input_ids": ids, "attention_mask": torch.ones_like(ids)}


@app.get("/v1/models")
def list_models():
    return {
        "object": "list",
        "data": [{"id": served_name, "object": "model", "owned_by": "whiterabbit"}],
    }


@app.post("/v1/chat/completions")
def chat_completions(req: dict):
    try:
        prompt = ""
        latest_user = ""
        for message in req.get("messages", []):
            role = str(message.get("role", "user")).upper()
            content = message.get("content", "")
            if isinstance(content, list):
                content = " ".join(
                    str(part.get("text", "")) if isinstance(part, dict) else str(part)
                    for part in content
                )
            if role in ("USER", "HUMAN", "ASSISTANT", "SYSTEM"):
                prompt += f"{role}: {content}\n"
            if role == "TOOL":
                prompt += f"USER: Result from {message.get('name', 'tool')}: {str(content)[:2500]}\n"
            if role == "ASSISTANT" and message.get("tool_calls"):
                for call in message["tool_calls"]:
                    function = call.get("function") or {}
                    prompt += f"ASSISTANT: Called {function.get('name', 'tool')}({function.get('arguments', '')[:900]})\n"
            if role in ("USER", "HUMAN"):
                latest_user = str(content)
        prompt += "ASSISTANT:"
        max_tokens = max(1, min(int(req.get("max_tokens") or 384), 1024))
        temperature = float(req.get("temperature", 0.65))
        do_sample = temperature > 0
        now = int(time.time())
        completion_id = f"chatcmpl-{time.time_ns()}"
        options = {
            "max_new_tokens": max_tokens,
            "do_sample": do_sample,
            "temperature": temperature if do_sample else 1.0,
            "top_p": float(req.get("top_p", 0.95)) if do_sample else 1.0,
            "top_k": 50,
            "repetition_penalty": 1.08,
            "pad_token_id": tokenizer.eos_token_id,
        }

        def chunk(delta, finish_reason=None):
            return f"data: {json.dumps({'id': completion_id, 'object': 'chat.completion.chunk', 'created': now, 'model': served_name, 'choices': [{'index': 0, 'delta': delta, 'finish_reason': finish_reason}]})}\n\n"

        if req.get("stream", False):
            def events():
                yield chunk({"role": "assistant"})
                with generation_lock:
                    inputs = build_inputs(prompt, latest_user)
                    generate_options = {**options, "stopping_criteria": StoppingCriteriaList([StopOnDialogue(inputs["input_ids"].shape[1])])}
                    streamer = TextIteratorStreamer(tokenizer, skip_prompt=True, skip_special_tokens=True, timeout=300)
                    errors = []

                    def generate():
                        try:
                            with torch.inference_mode():
                                model.generate(**inputs, streamer=streamer, **generate_options)
                        except Exception as exc:
                            errors.append(exc)
                            streamer.end()

                    worker = threading.Thread(target=generate, daemon=True)
                    worker.start()
                    pending = ""
                    stopped_at_dialogue = False
                    try:
                        for part in streamer:
                            pending += part
                            clean = clean_dialogue_tail(pending)
                            if clean != pending.strip():
                                if clean:
                                    yield chunk({"content": clean})
                                stopped_at_dialogue = True
                                break
                            if len(pending) > 24:
                                yield chunk({"content": pending[:-24]})
                                pending = pending[-24:]
                    finally:
                        worker.join()
                        del inputs, streamer
                    if errors:
                        print("Generation failed:", repr(errors[0]), flush=True)
                        yield f"data: {json.dumps({'error': {'message': str(errors[0]), 'type': 'server_error'}})}\n\n"
                        return
                    if pending and not stopped_at_dialogue:
                        yield chunk({"content": pending})
                yield chunk({}, "stop")
                yield "data: [DONE]\n\n"

            return StreamingResponse(events(), media_type="text/event-stream", headers={"Cache-Control": "no-cache", "X-Accel-Buffering": "no"})

        with generation_lock, torch.inference_mode():
            inputs = build_inputs(prompt, latest_user)
            input_length = inputs["input_ids"].shape[1]
            outputs = model.generate(
                **inputs,
                stopping_criteria=StoppingCriteriaList([StopOnDialogue(input_length)]),
                **options,
            )
            tokens = outputs[0][input_length:]
            text = clean_dialogue_tail(tokenizer.decode(tokens, skip_special_tokens=True))
            completion_tokens = len(tokens)
            del inputs, outputs, tokens

        return {
            "id": completion_id,
            "object": "chat.completion",
            "created": now,
            "model": served_name,
            "choices": [{"index": 0, "message": {"role": "assistant", "content": text}, "finish_reason": "stop"}],
            "usage": {
                "prompt_tokens": input_length,
                "completion_tokens": completion_tokens,
                "total_tokens": input_length + completion_tokens,
            },
        }
    except Exception as exc:
        torch.cuda.empty_cache()
        gc.collect()
        import traceback
        traceback.print_exc()
        return JSONResponse(status_code=500, content={"error": {"message": str(exc), "type": "server_error"}})


def tool_instruction(tools):
    definitions = []
    for entry in tools[:16]:
        function = entry.get("function") or {}
        name = str(function.get("name", ""))
        if not name:
            continue
        parameters = function.get("parameters") or {}
        required = set(parameters.get("required") or [])
        args = []
        for key, value in list((parameters.get("properties") or {}).items())[:12]:
            if not isinstance(value, dict):
                continue
            desc = str(value.get("description", "")).replace("\n", " ")[:75]
            arg_type = value.get("type", "value")
            args.append(f"{key}: {arg_type}{' required' if key in required else ''} {desc}")
        definitions.append(f"{name}: {str(function.get('description', ''))[:120]} ({'; '.join(args)})")
    return (
        "Available tools:\n" + "\n".join(definitions) + "\n"
        "To use a tool, output ONLY <tool_call>{\"name\":\"tool_name\",\"arguments\":{...}}</tool_call> "
        "with valid JSON arguments. Use exactly one tool at a time. "
        "After a tool result, either call another tool or answer the request. "
        "If no tool is needed, answer normally; never invent a tool result."
    )


def parse_tool_call(text, tools):
    allowed = {entry.get("function", {}).get("name") for entry in tools}
    candidates = re.findall(r"<tool_call>\s*(\{.*?\})\s*</tool_call>", text, re.DOTALL)
    if not candidates:
        candidates = re.findall(r"```(?:json)?\s*(\{.*?\})\s*```", text, re.DOTALL)
    candidates.append(text.strip())
    decoder = json.JSONDecoder()
    for candidate in candidates:
        start = candidate.find("{")
        if start < 0:
            continue
        try:
            parsed, _ = decoder.raw_decode(candidate[start:])
        except (ValueError, TypeError):
            continue
        if not isinstance(parsed, dict) or parsed.get("name") not in allowed:
            continue
        arguments = parsed.get("arguments")
        if isinstance(arguments, str):
            try:
                arguments = json.loads(arguments)
            except ValueError:
                continue
        if isinstance(arguments, dict):
            return parsed["name"], json.dumps(arguments, ensure_ascii=False)
    return None


_plain_chat_completions = chat_completions
app.router.routes[:] = [r for r in app.router.routes if getattr(r, "path", "") != "/v1/chat/completions"]


@app.post("/v1/chat/completions")
def chat_completions_with_tools(req: dict):
    tools = req.get("tools") or []
    choice = req.get("tool_choice", "auto")
    if not tools or choice == "none":
        return _plain_chat_completions(req)

    prepared = dict(req)
    prepared["stream"] = False
    prepared["messages"] = list(req.get("messages") or []) + [
        {"role": "system", "content": tool_instruction(tools)}
    ]
    prepared["max_tokens"] = max(256, min(512, int(req.get("max_tokens") or 512)))
    prepared["temperature"] = 0.2
    response = _plain_chat_completions(prepared)
    if isinstance(response, JSONResponse):
        return response
    answer = response["choices"][0]["message"]["content"]
    parsed = parse_tool_call(answer, tools)
    message = {"role": "assistant", "content": answer}
    finish_reason = "stop"
    if parsed:
        name, arguments = parsed
        message = {
            "role": "assistant", "content": None,
            "tool_calls": [{
                "id": "call_" + uuid.uuid4().hex[:24], "type": "function",
                "function": {"name": name, "arguments": arguments},
            }],
        }
        finish_reason = "tool_calls"
    response["choices"][0]["message"] = message
    response["choices"][0]["finish_reason"] = finish_reason
    if not req.get("stream"):
        return response

    def events():
        base = {key: response[key] for key in ("id", "created", "model")}
        base["object"] = "chat.completion.chunk"

        def chunk(delta, reason=None):
            payload = {**base, "choices": [{"index": 0, "delta": delta, "finish_reason": reason}]}
            return f"data: {json.dumps(payload)}\n\n"

        yield chunk({"role": "assistant"})
        if parsed:
            call = message["tool_calls"][0]
            yield chunk({"tool_calls": [{"index": 0, **call}]})
        elif answer:
            yield chunk({"content": answer})
        yield chunk({}, finish_reason)
        yield "data: [DONE]\n\n"

    return StreamingResponse(events(), media_type="text/event-stream")


server_thread = threading.Thread(
    target=lambda: uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning"),
    daemon=True,
)
server_thread.start()
time.sleep(2)

print("Starting Cloudflare tunnel...", flush=True)
tunnel = subprocess.Popen(
    ["cloudflared", "tunnel", "--url", "http://127.0.0.1:8000"],
    stderr=subprocess.PIPE,
    text=True,
)
for line in tunnel.stderr:
    match = re.search(r"https://[a-zA-Z0-9-]+\.trycloudflare\.com", line)
    if match:
        api_base = match.group(0) + "/v1"
        with open("/kaggle/working/tunnel_url.txt", "w") as target:
            target.write(api_base + "\n")
        print("API BASE:", api_base, flush=True)
        break
tunnel.wait()
````
